I created a copy to fall back in if anything goes wrong, first drop the rows with martial status as ? since its only 1 rows as we can see by running a boolean and .sum

```python
(df['Marital status'] == '?').sum()
```

```python
df_copy = df_copy[df_copy['Marital status'] != '?']
```

```python
(df['Nationality'] == '?').sum()
```

nationality only has 1 so i drop that too

```python
df_copy = df_copy[df_copy['Nationality'] != '?']
```

Fathers occupation is set to five instead of 5, so i just replace it

```python
df_copy["Father's occupation"] = df_copy["Father's occupation"].replace('Five', '5')
```

```python
(df_copy['Tuition fees up to date'] == '?').sum()
```

again tutition fees only has 1 row with a ? so i drop that

```python
df_copy = df_copy[df_copy['Tuition fees up to date'] != '?']
```

age at enrollment and curricular untis 1st sem credited only have 1 row each wrong so i drop them

```python
df_copy = df_copy[df_copy['Age at enrollment'] != 'UnKnown']
df_copy = df_copy[df_copy['Curricular units 1st sem (credited)'] != 'Na']
 
```

Now that its all cleaned up i will turn the string variables into integers because we cant do any math or analysis on them if they stay as strings

```python
df_copy['Marital status'] = df_copy['Marital status'].astype(int)
df_copy['Nationality'] = df_copy['Nationality'].astype(int)
df_copy["Father's occupation"] = df_copy["Father's occupation"].astype(int)
df_copy['Displaced'] = df_copy['Displaced'].astype(int)
df_copy['Tuition fees up to date'] = df_copy['Tuition fees up to date'].astype(int)
df_copy['Age at enrollment'] = df_copy['Age at enrollment'].astype(int)
df_copy['Curricular units 1st sem (credited)'] = df_copy['Curricular units 1st sem (credited)'].astype(int)
```

Next i will create EDA visualisations using matplot and Seaborn

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.countplot(x='Target', data=df)
plt.title('Student Outcomes')
plt.show()
```

We can see multiple Dropout , as Drop and Dropout, so i will clean this for better visualisation and because without this step the model would treat them as seperate classes. This is why EDA is useful , because without it I might not have spotted this inconsitency 

```python
df_copy['Target'].value_counts()
```

```python
df_copy['Target'] = df_copy['Target'].replace({
    'Grad': 'Graduate',
    'graduate': 'Graduate',
    'Drop': 'Dropout',
    'dropout': 'Dropout'
})
```

```python
df_copy['Target'].unique()
```

```python
sns.countplot(x='Target', data=df_copy)
plt.title('Student Outcomes')
plt.show()
```

Age vs outcome

```python
sns.boxplot(x='Target', y='Age at enrollment', data=df_copy)
plt.title('Age at Enrollment by Outcome')
plt.show()
```

I used the above because it shows that Dropouts are more widely spread, dropouts are older. The outliers are real and not errors as they are mature students so i wont remove them

```python
sns.countplot(x='Tuition fees up to date', hue='Target', data=df_copy)
plt.title('Tuition Fees Up to Date by Outcome')
plt.show()
```

Im seeing when it comes to not having tuition fees up to date, there is a lot of dropouts, so this is a major link for drop outs. But we are also seeing alot more that have them up to date are also dropping out. So we cant say for certain as other factors may have been in play here

```python
sns.scatterplot(x='Curricular units 1st sem (approved)',
                y='Curricular units 2nd sem (approved)',
                hue='Target', data=df_copy, alpha=0.5)
plt.title('Units Approved: 1st vs 2nd Semester')
plt.show()
```

A scatterplot was used to examine between units approved in the 1st and 2nd semester
There is a strong positive between students who pass many units in semester 1 general pass semster 2
