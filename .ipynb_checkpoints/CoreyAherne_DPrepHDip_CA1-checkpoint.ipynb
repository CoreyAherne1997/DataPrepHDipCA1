Now that its all cleaned up i will turn the string variables into integers because we cant do any math or analysis on them if they stay as strings

```python
df_copy['Marital status'] = df_copy['Marital status'].astype(int)
df_copy['Nationality'] = df_copy['Nationality'].astype(int)
df_copy["Father's occupation"] = df_copy["Father's occupation"].astype(int)
df_copy['Displaced'] = df_copy['Displaced'].astype(int)
df_copy['Tuition fees up to date'] = df_copy['Tuition fees up to date'].astype(int)
df_copy['Age at enrollment'] = df_copy['Age at enrollment'].astype(int)
df_copy['Curricular units 1st sem (credited)'] = df_copy['Curricular units 1st sem (credited)'].astype(int)
```

next we train the data by splitting it 

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

X_train, X_test, y_train, y_test = train_test_split(X_scaler, y, test_size=0.2, random_state=2, stratify=y)
Xp_train, Xp_test, yp_train, yp_test = train_test_split(X_pca, y, test_size=0.2, random_state=2, stratify=y)
```

Model on original data

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
pred = model.predict(X_test)

print('Original accuracy:', accuracy_score(y_test, pred))
print(classification_report(y_test, pred))
```

Model on PCA data

```python
model_pca = LogisticRegression(max_iter=1000)
model_pca.fit(Xp_train, yp_train)
pred_pca = model_pca.predict(Xp_test)

print('PCA accuracy:', accuracy_score(yp_test, pred_pca))
print(classification_report(yp_test, pred_pca))
```

Not that much difference between the 2, which is to be expected since it only went form 36 to 33 to get 99.5%.Precision, recall and accuracy are about the same with only 1 or 2 variance between them
