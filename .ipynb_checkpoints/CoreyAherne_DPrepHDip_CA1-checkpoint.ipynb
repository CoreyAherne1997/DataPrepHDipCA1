```python
df_copy = df.copy()
```

I created a copy to fall back in if anything goes wrong, first drop the rows with martial status as ? since its only 1 rows as we can see by running a boolean and .sum

```python
(df['Marital status'] == '?').sum()
```

```python
df_copy = df_copy[df_copy['Marital status'] != '?']
```

```python
(df['Nationality'] == '?').sum()
```

nationality only has 1 so i drop that too

```python
df_copy = df_copy[df_copy['Nationality'] != '?']
```

Fathers occupation is set to five instead of 5, so i just replace it

```python
df_copy["Father's occupation"] = df_copy["Father's occupation"].replace('Five', '5')
```

```python
(df_copy['Tuition fees up to date'] == '?').sum()
```

again tutition fees only has 1 row with a ? so i drop that

```python
df_copy = df_copy[df_copy['Tuition fees up to date'] != '?']
```

age at enrollment and curricular untis 1st sem credited only have 1 row each wrong so i drop them

```python
df_copy = df_copy[df_copy['Age at enrollment'] != 'UnKnown']
df_copy = df_copy[df_copy['Curricular units 1st sem (credited)'] != 'Na']
 
```

Now that its all cleaned up i will turn the string variables into integers because we cant do any math or analysis on them if they stay as strings

```python
df_copy['Marital status'] = df_copy['Marital status'].astype(int)
df_copy['Nationality'] = df_copy['Nationality'].astype(int)
df_copy["Father's occupation"] = df_copy["Father's occupation"].astype(int)
df_copy['Displaced'] = df_copy['Displaced'].astype(int)
df_copy['Tuition fees up to date'] = df_copy['Tuition fees up to date'].astype(int)
df_copy['Age at enrollment'] = df_copy['Age at enrollment'].astype(int)
df_copy['Curricular units 1st sem (credited)'] = df_copy['Curricular units 1st sem (credited)'].astype(int)
```
